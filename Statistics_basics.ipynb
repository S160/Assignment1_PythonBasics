Ans 1: Descriptive Statistics:- It just summarizes and describes data that you already have. It's purpose is to give a clear picture of what the data looks like. Tools like the mean (average), median, mode, range, standard deviation, charts, tables, etc. are used in descriptive statistics.

Inferential Statistics:-  It uses data from a sample to make conclusions or predictions about a larger population. It's purpose is to go beyond the data at hand and infer something more general. Tools like the Hypothesis testing, confidence intervals, regression, correlation, probability etc. are much used in Inferential statistics

Ans 2: Sampling is the process of selecting a subset (sample) from a larger group (population) so that we can study the sample and make conclusions about the whole population.

Population = the entire group you are interested in (e.g., all students in a university).

Sample = a smaller part of the population (e.g., 200 students chosen from the university).
We use sampling because it's often impossible, too costly, or time-consuming to study the entire population.

Random Sampling:- Every member of the population has an equal chance of being selected. Selection is done purely by chance (like using a lottery, random number generator, or drawing names).
For example:- If you want to survey 1,000 students from a university of 20,000, you randomly select 1,000 names without looking at gender, department, or year.
It's simple unbiased and fair.

Stratified Sampling:- In this, the population is divided into subgroups (strata) based on some characteristic (like gender, age, department). hen, samples are taken from each subgroup, often proportionally.
For example, In the same university, you first divide students by faculty (Engineering, Arts, Commerce, Science). Then you randomly select students from each faculty in proportion to their size.

Ans 3: Mean:- The mean is the sum of all the values in a dataset divided by the total number of values.
        Mean = Sum of all observations/ Number of Observations
Median:- The median is the middle value of the dataset when arranged in ascending (or descending) order.
Mode:- The mode is the value that occurs most frequently in the dataset.

The importance of the measures of central tendency are:-
1. Summarization: They give a single representative value that summarizes a whole dataset.
2. Comparison: Allow comparison between different groups or datasets.
3. Decision-Making: Widely used in business, economics, psychology, and research for analysis and predictions.
4. It is very important for understanding the distribution. Mean is very useful for evenly distributed data. Median is more reliable when data has extreme values.

Ans 4: Skewness measures the asymmetry of the data distribution around its mean. Kurtosis measures the tailedness or how heavy/light the tails of a distribution are compared to a normal distribution.
Positive skew is the Right-skewed, which means that ail is longer on the right side.
Mean > Median > Mode.

For example:- Income data (a few people earn extremely high incomes).

```python
import statistics as stats

# Given list of numbers
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Calculating mean
mean_val = stats.mean(numbers)

# Calculating median
median_val = stats.median(numbers)

# Calculating mode
mode_val = stats.mode(numbers)

# Display results
print("Mean:", mean_val)
print("Median:", median_val)
print("Mode:", mode_val)
```

```python
import numpy as np

# Given datasets
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Convert to numpy arrays
x = np.array(list_x)
y = np.array(list_y)

# Covariance matrix
cov_matrix = np.cov(x, y, bias=False)  # by default gives sample covariance

# Extract covariance between x and y
cov_xy = cov_matrix[0, 1]

# Correlation coefficient
corr_xy = np.corrcoef(x, y)[0, 1]

# Display results
print("Covariance between X and Y:", cov_xy)
print("Correlation Coefficient between X and Y:", corr_xy)
```

```python
#Python code to draw a boxplot for the numeric list, data= [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 25, 29, 35]
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Create Boxplot
sns.boxplot(data=data)
plt.title("Boxplot of Given Data")
plt.show()

# Detect Outliers using IQR
Q1 = np.percentile(data, 25)   # First quartile
Q3 = np.percentile(data, 75)   # Third quartile
IQR = Q3 - Q1                  # Interquartile Range

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

Ans 8: Covariance measures how two variables change together.
* If positive, advertising spend and daily sales increase together.
* If negative, one goes up when the other goes down.
But covariation does not tell us about the strength of the relationship.

Correlation is the Standardized version of covariance (ranges between -1 and +1).
1. +1 → perfect positive linear relationship

2. -1 → perfect negative linear relationship

3.  0 → no linear relationship.

```python
import numpy as np

# Data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

# Calculate covariance matrix
cov_matrix = np.cov(advertising_spend, daily_sales, ddof=0)
covariance = cov_matrix[0, 1]

# Calculate correlation coefficient
correlation = np.corrcoef(advertising_spend, daily_sales)[0, 1]

print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)
```

Ans 9: We will do the EDA (Exploratory Data Analysis) here.
The summary statistics use depends on the following:

* Mean:- Tells the central tendency of satisfaction (overall score level).
* Median:- Helps understand the middle value, especially if the distribution is skewed.
* Mode:- Shows the most common satisfaction score (important in surveys).
* Standard Deviation (SD) & Variance:- It indicates how spread out the scores are.
* Minimum & Maximum:- To know the range of satisfaction (worst vs best feedback).

Visualizations to be used are:
1. Histogram
2. Boxplot
3. Bar Plot of Frequencies

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Data
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

# Summary Statistics
mean_score = np.mean(survey_scores)
median_score = np.median(survey_scores)
mode_score = max(set(survey_scores), key=survey_scores.count)
std_dev = np.std(survey_scores)
min_score = min(survey_scores)
max_score = max(survey_scores)

print("Mean:", mean_score)
print("Median:", median_score)
print("Mode:", mode_score)
print("Standard Deviation:", std_dev)
print("Range:", (min_score, max_score))

# Histogram
plt.hist(survey_scores, bins=6, edgecolor="black")
plt.title("Distribution of Customer Satisfaction Scores")
plt.xlabel("Survey Score (1-10)")
plt.ylabel("Frequency")
plt.show()
```
