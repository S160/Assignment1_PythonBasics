Ans 1: Dimensionality Reduction is the process of reducing the number of input features (dimensions) in a dataset while preserving as much important information as possible.
Common methods are:
1. PCA (Principal Component Analysis)
2. t-SNE
3. LDA (Linear Discriminant Analysis)
4. Autoencoders

**Importance In Machine Learning:-**
* Reduces Overfitting:- High-dimensional data can cause models to memorize noise. Reducing dimensions removes irrelevant or redundant features → better generalization
* Improves Model Performance:- Less data means: faster training, lower memory usage, simpler models.
* Handles the curse of Dimensionality:- In very high dimensions, distance-based algorithms (KNN, clustering) break down. Reducing dimensions restores meaningful distances.
* Helps in Visualization:- You can compress data to 2D and 3D. This helps to visualize clusters, classes, and patterns.
* Removes Multicollinearity:- PCA converts correlated features into uncorrelated principal components, improving model stability.





Ans 2: The three common dimensionality reduction techniques are:
1. PCA (Principal Component Analysis):- It is a linear and unsupervised method that converts original features into new components. It keeps components with maximum variance. It is one of the most widely used reduction techniques best used for numeric and continuous data (for example, images, sensors, tabular data).
2. t-SNE:- It is a non-linear and unsupervised method which is best for visualization (2D or 3D) or high-dimensional messy data. It is very good for clustering visualizations as it uses probability distributions to represent similarities and shows clear clusters. It preserves local structure: points that are close in high-dim space stay close in low-dim space.
3. LDA (Linear Discriminant Analysis):- It is a supervised technique which maximizes the class seperation by finding a new axes that maximizes distance between classes and minimizes variance inside each class. It is used in classification tasks like face recognition and medical diagnosis.

Ans 3: Clustering is an unsupervised learning technique where the algorithm groups data points based on similarity-without using labels. Data points in the same cluster are more similar to each other than to other points in other clusters.

We don't tell the model what groups exists, it discovers them on its own. For example, grouping customers by shopping behaviour, growing flowers by petal measurements etc.

Three popular clustering algorithms are:-
1. K-means clustering
2. Heirarchical Clustering
3. DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

Ans 4: Anomaly Detection is the process of identifying data points or events that are unusual, rare, or do not follow the normal pattern.
These unusual points are called:-
* Anomalies
* Outliers and
* Deviations

In other words, "Anything that looks different from the rest is an anomaly". Some of the examples of anomaly detection are:-
1. **Fraud Detection**:- This involves strange credit card transaction, buying something at an unusual location & very high-value unexpected purchase.
Banks catch fraud using anomaly detection.
2. **Network & Cybersecurity**:- This involves suspicious login attempts, unusual traffic from a device and malware activity. It helps prevent hacking and security breaches.
3. **Medical & Health Monitoring**:- This includes abnormal heartbeat, strange patterns in medical reports and sudden spikes in sugar levels. These anomalies can warn about health risks.
4. **Industrial Monitoring**:- Machines behave differently and sometimes an unexpected temperature rise. Sensor readings are out of normal range. These anomalies are used to prevent equipment failure later.
5. **Data Cleaning**:- Anomalies help identify wrong entries, missing sensor values and human typing mistakes.

Ans 5: The three types of Anomaly Detection techniques are:-
1. Statistical Methods:- Look for values that deviate strongly from a statistical distribution. Detects anomalies by identifying outliers far from mean/median. Examples:- Z-score, Gaussian models, Boxplots.

2. Distance-Based Methods:- Measures how far a point is from others and if a point is far from all neighbors it is an anomaly. For example, KNN-based anomaly detection, Distance thresholding, etc.

3. Clustering-Based Methods:- Assume anomalies do not belong to any cluster or lie in very small clusters. For example:- DBSCAN (points in low-density areas = anomalies), K-Means (points far from cluster centers = anomalies)

Ans 6: Time-Series Analysis is the study of data points collected over time at regular intervals (daily, monthly, yearly, etc.) to understand patterns, trends, and future behavior.
It helps in forecasting things like:
*  Stock prices
*  Whether
*  Sales
*  Traffic
*  Electricity Demand

The two key components of Time-Series data are:-
1. **Trend**:- A long-term upward or downward movement in the data. For example, company's sales increasing every year
2. **Seasonality**:- Regular, repeating patterns within fixed intervals. For example, Ice-cream sales go up every summer.


Ans 7: Seasonality refers to the repeating patterns that occur at regular, fixed intervals wheras cyclic behaviour refers to the patterns that rise & fall over long, irregular intervals.

The interval of seasonality is known and consistent, for example:- every day, week, month, quarter,year whereas the intervals of cyclic behaviour have no fixed or predictable periods.

The cause of seasonality is calender-related or environmental factors whereas the cause of cyclic behaviour is mainly economic, social or external factors.

Examples of Seasonality:- Ice-cream sales increasing every year, Online shopping spikes during weekends, Electricity demand rising every evening. Examples of Cyclic behaviour:- Economic recessions, Stock market cycles, Real estate booms & slowdowns

```python
# K-Means Clustering Example
# --------------------------

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# 1. Generate sample dataset
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

# 2. Train K-Means model
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

# 3. Get cluster labels
labels = kmeans.labels_

# 4. Print cluster centers
print("Cluster Centers:\n", kmeans.cluster_centers_)

# 5. Plot results
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(kmeans.cluster_centers_[:, 0],
            kmeans.cluster_centers_[:, 1],
            s=200, marker='X')
plt.title("K-Means Clustering")
plt.show()
```

Ans 9: Inheritance is an Object-Oriented Programming (OOP) concept where a new class (called a child class or subclass) can reuse, extend, or modify the behavior of an existing class (called a parent class or base class)

It allows you to:-
*  avoid writing duplicate codes
*  build heirarchical relationships
*  create more organised & maintainable programs



```python
# Parent Class
class Animal:
    def sound(self):
        return "Some generic animal sound"

# Child Class (inherits from Animal)
class Dog(Animal):
    def sound(self):
        return "Bark"

# Another Child Class
class Cat(Animal):
    def sound(self):
        return "Meow"

# Using the classes
a = Animal()
d = Dog()
c = Cat()

print(a.sound())  # Output: Some generic animal sound
print(d.sound())  # Output: Bark
print(c.sound())  # Output: Meow
```

Ans 10: Time series analysis can detect anomalies by studying how a variable behaves over time & identifying points that don't follow the expected pattern.
1. **Using Forecast Model**:- You can forecast the next values uusing models like:
*  ARIMA/SARIMA
*  Exponential Smoothing
*  LSTM/RNN models

Then compare:-

      Actual value-Predicted value

If the difference is too large, that point is flagged as an Anomaly. For example, if your system predicts power consumption should be 200-220 KW but it suddenly jumps to 310 KW or more, then it is an Anomaly.

2. **Using Statistical Thresholds**:- From the time series compute:-
*  Mean
*  Standard Deviation
*  Moving Average
*  Moving Variance

Mark any point that lies outside the normal range (eg- >3 standard deviation) as Anomaly

3. **Using Decomposition**:- TIme series decomposition splits the signal into:-
* Trend
* Seasonality
* Residual

This residual part shows irregular behaviour. If residual spike is too large its an Anomaly.

This works because time series naturally captures patterns, trends, cycles, seasonality. So the sudden deviations become very easy to spot.
