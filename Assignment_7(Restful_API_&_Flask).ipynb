Ans 1: A RESTful API (Representational State Transfer API) is a way for computer systems to communicate with each other over the web using HTTP methods.

It follows the REST architecture, which is based on a set of principles that make APIs simple, scalable, and easy to use.

1. Client-Server Model:- The client (like a web app or mobile app) sends requests, and the server responds with resources.
2. Stateless:- Each request is independent. The server does not remember the client’s state between requests.

  Example: If you want to fetch a user’s data, you must send authentication details each time.
3. Resources & URIs:-Everything is treated as a resource (user, order, product, etc.). Resources are identified by URIs (Uniform Resource Identifiers).

  Example: GET /users → fetch all users

  GET /users/1 → fetch user with ID=1
4. HTTP Methods (mapped to CRUD operations):
  
  GET → Read data

  POST → Create new data

  PUT → Update existing data

  DELETE → Remove data
5. Representation of Resources:- Data is usually sent in JSON (sometimes XML).

Flask-SQLAlchemy is a Flask extension that integrates SQLAlchemy (Python's most popular ORM- Object Relational Mapper) with Flask. It allows you to work with databases using Python classes and objects instead of writing raw SQL queries. In simple words, it helps you connect Flask apps to databases (like SQLite, MySQL, PostgreSQL) and interact with them in a Pythonic way.

The roles of the Flask SQLAlchemy are discussed as follows
 1. **Database Connection Management**:- Configures your database connection using a single line in app.config. It can connect to SQLite, MySQL, PostgreSQL, Oracle, MS SQL Server etc.
 For example,

```python
app.config['SQLALCHEMY_DATABASE_URI'] = 'sqlite:///mydb.sqlite'
```

For example:-

```python
from flask_sqlalchemy import SQLAlchemy
db = SQLAlchemy(app)

class User(db.Model):
    id = db.Column(db.Integer, primary_key=True)
    name = db.Column(db.String(50), nullable=False)
```

Ans 14: Steps to create a RESTful APIs Endpoint using Flask
1. Firstly, import Flask & create the app:-
   
   Code:-

```python
from flask import Flask, request, jsonify
app = Flask(__name__)
```

2. Defining a sample data store (in memory for demo)

   Code:-
   

```python
# Fake in-memory "database"
users = [
    {"id": 1, "name": "Shreya", "age": 24},
    {"id": 2, "name": "Arjun", "age": 28}
]
```

3. Create RESTful API Endpoint:-

    a) GET – Retrieve all users
    
   Code:-

```python
@app.route("/users", methods=["GET"])
def get_users():
    return jsonify(users)
```

b) GET by ID – Retrieve a single user
   
   Code:-

```python
@app.route("/users/<int:user_id>", methods=["GET"])
def get_user(user_id):
    user = next((u for u in users if u["id"] == user_id), None)
    if user:
        return jsonify(user)
    return jsonify({"error": "User not found"}), 404
```

c) POST - Create a new user
   
   Code:-

```python
@app.route("/users", methods=["POST"])
def create_user():
    data = request.json
    new_user = {
        "id": len(users) + 1,
        "name": data["name"],
        "age": data["age"]
    }
    users.append(new_user)
    return jsonify(new_user), 201
```

d) PUT- Update a user

   Code:-

```python
@app.route("/users/<int:user_id>", methods=["PUT"])
def update_user(user_id):
    data = request.json
    user = next((u for u in users if u["id"] == user_id), None)
    if not user:
        return jsonify({"error": "User not found"}), 404
    user.update(data)
    return jsonify(user)
```

e) DELETE:- Delete a user

   Code:-

```python
@app.route("/users/<int:user_id>", methods=["DELETE"])
def delete_user(user_id):
    global users
    users = [u for u in users if u["id"] != user_id]
    return jsonify({"message": "User deleted"})
```

4. Run the Flask app:-

   Code:-

```python
if __name__ == "__main__":
    app.run(debug=True)
```

Ans 18: An API Specification is a blueprint or contract that describes:

1. What endpoints (URLs) the API provides

2. What HTTP methods (GET, POST, PUT, DELETE, etc.) are allowed

3. What input data (query params, request body) is expected

4. What responses (status codes, JSON structure, error messages) will be returned

Think of it like an instruction manual for developers who want to use or build the API.

It helps building a Flask API by the following:-
1. Clear communication- Developers (backend + frontend + QA) know exactly what to expect from each API which reduces misunderstandings.
2. Faster development- Frontend team can mock API responses before the backend is finished. Backend team can implement APIs knowing what format is expected.
3. Easy validation- Tools like Flask-RESTX, Flask-Smorest, or Marshmallow can validate requests and responses against the spec automatically.
4. Documentation for free- API specs are often written in OpenAPI/Swagger format. With a spec, Flask APIs can auto-generate interactive documentation (Swagger UI, Redoc).
5. Easier testing:- Testers can check API responses against the spec to verify correctness.

Ans 19: HTTP status codes are 3-digit numbers returned by a web server (including a Flask API) in response to a client's request.

1. 1xx - Informational
2. 2xx - Success
3. 3xx - Redirection
4. 4xx - Client Errors
5. 5xx - Server Errors

They are importent in Flask API because of the following reasons:-
1. Communication between client & server:- Clients (like browsers, mobile apps, or frontend apps) rely on status codes to know what happened.

2. Error handling:- Instead of just saying "error", status codes give clear context:
400 → Client messed up the request
500 → Server crashed

3. RESTful API best practices:- REST APIs should follow standard status codes so developers know how to handle responses.

4. Debugging & logging:- Makes it easier to track issues in logs.

```python
#Code to handle errors in Flask
from flask import Flask, render_template

app = Flask(__name__)

@app.route("/")
def home():
    return "Welcome to the Home Page!"

# Example route to trigger error
@app.route("/cause-error")
def cause_error():
    # This will raise a 500 error
    1 / 0

# Handle 404 (Page Not Found)
@app.errorhandler(404)
def page_not_found(e):
    return render_template("404.html"), 404


#templates/404.html
<!DOCTYPE html>
<html>
<head>
    <title>Page Not Found</title>
</head>
<body>
    <h1>❌ 404 Error</h1>
    <p>Sorry, the page you are looking for does not exist.</p>
    <a href="{{ url_for('home') }}">Go Back Home</a>
</body>
</html>
```

```python
#Code to structure a Flask app using Blueprints.

#Program Structure
myapp/
│
├── app.py              # Main entry point
├── home/
│   ├── __init__.py     # Make 'home' a package
│   └── routes.py       # Routes for home blueprint
└── auth/
    ├── __init__.py
    └── routes.py       # Routes for auth blueprint

#app.py(Main app)
from flask import Flask
from home.routes import home_bp
from auth.routes import auth_bp

app = Flask(__name__)

# Register Blueprints
app.register_blueprint(home_bp)
app.register_blueprint(auth_bp, url_prefix="/auth")

if __name__ == "__main__":
    app.run(debug=True)

#home/routes.py (Home Blueprint)
from flask import Blueprint

# Create Blueprint
home_bp = Blueprint("home", __name__)

@home_bp.route("/")
def index():
    return "Welcome to the Home Page!"

#auth/routes.py (Auth Blueprint)
from flask import Blueprint

# Create Blueprint
auth_bp = Blueprint("auth", __name__)

@auth_bp.route("/login")
def login():
    return "Login Page"

@auth_bp.route("/logout")
def logout():
    return "Logout Page"
```

```python
#Code to define a custom Jinja filter in Flask
from flask import Flask, render_template

app = Flask(__name__)

# Custom filter: reverse a string
def reverse_string(s):
    return s[::-1]

# Register the filter with Jinja
app.jinja_env.filters['reverse'] = reverse_string

@app.route("/")
def home():
    name = "Flask"
    return render_template("index.html", name=name)

if __name__ == "__main__":
    app.run(debug=True)

#templates/index.html
<!DOCTYPE html>
<html>
<head>
    <title>Custom Filter Example</title>
</head>
<body>
    <h1>Original: {{ name }}</h1>
    <h2>Reversed (using custom filter): {{ name | reverse }}</h2>
</body>
</html>
```

```python
#Code to rredirect with Query parameters
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

@app.route('/')
def home():
    # Redirect to /welcome with query parameters
    return redirect(url_for('welcome', name="Shreya", age=24))

@app.route('/welcome')
def welcome():
    name = request.args.get('name')
    age = request.args.get('age')
    return f"Welcome {name}, age {age}!"

if __name__ == '__main__':
    app.run(debug=True)
```

Ans 14: Code to return JSON responses in Flask

   Example 1:- Returning JSON with jsonify

```python
from flask import Flask, jsonify

app = Flask(__name__)

@app.route('/api/data')
def get_data():
    data = {
        "name": "Shreya",
        "age": 24,
        "city": "Kota"
    }
    return jsonify(data)

if __name__ == "__main__":
    app.run(debug=True)
```

Example 2:-JSON with Custom Status Code

```python
@app.route('/api/error')
def error_response():
    return jsonify({"error": "Something went wrong!"}), 400
```

```python
#Code to capture URL parameters in Flask
from flask import Flask

app = Flask(__name__)

# Capture one parameter
@app.route('/user/<username>')
def show_user(username):
    return f"Hello, {username}!"

# Capture multiple parameters
@app.route('/post/<int:post_id>')
def show_post(post_id):
    return f"Post ID: {post_id}"

if __name__ == "__main__":
    app.run(debug=True)
```
