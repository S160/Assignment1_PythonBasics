Ans 1: AI (Artificial Intelligence) is the broadest concept—it refers to creating systems or machines that can perform tasks that normally require human intelligence. It includes reasoning, problem-solving, learning, understanding language, vision, and decision-making. For example:- Chatbots.

ML (Machine Learning):It is a subset of AI that focuses on building algorithms that can learn from data and improve their performance over time without being explicitly programmed. It uses statistical methods to find patterns and make predictions. For example:- predicting prices of houses

DL (Deep Learning): It ia a subset of ML that uses artificial neural networks (inspired by the human brain) with many layers ("deep") to model complex patterns in data. It handles unstructured data like images, sounds and also predicts the figuures and prices. For example:- Face recognition, voice assistants (Alexa, Siri).

Data Science (DS):- It is an interdisciplinary field that focuses on extracting insights and knowledge from data using statistics, ML, visualization, and domain expertise. It has a scope in and covers data collection, cleaning, analysis, visualization, and communicating results for decision-making.

It is basically like a toolbox which uses AI, ML and DL techiniques to solve real-world problems.
For example: Analyzing customer churn, business dashboards, fraud detection reports.

Ans 2:  Machine Learning (ML) is usually classified into three main types and sometimes the additional fourth type which are as follows:-

1. Supervised Learning:- In this type of ML, the model is trained on a labeled dataset (i.e., input data with correct answers/outputs). The goal is to learn the mapping between inputs and outputs. It works by learning from the past examples to predict the future.
It performs two major tasks:- Regression (Predicting continuous values) and Classification (Predicting categories).

    For example:- Spam Detection E-mails

2. Unsupervised Learning: In this type of ML, the model is trained on unlabeled data (no predefined answers). The system tries to find hidden patterns, structures, or relationships in the data. It works naturally by finding what groups together.
It performs two major tasks:- Clustering (Grouping similar data points) and Dimensionallity Reduction (Simplify data by retaining meaning).

      For example:- Customer segmentation in marketing

3. Reinforcement Learning (RL):- In this type of ML, he model learns by interacting with an environment. It takes actions, receives feedback (rewards/penalties), and learns the best strategy (policy) to maximize long-term rewards.
It works by the method of Trial and Error with Feedback.

      For example:- Self-driving cars.
4. Semi-supervised Learning (Hybrid of Supervised + Unsupervised):- In this type, the system uses a small amount of labeled data plus a large amount of unlabeled data. Useful when labeling is expensive or difficult.

      For example:- Medical Diagnosis with X-rays


Ans 3: Overfitting:- It is defined as When a machine learning model learns the training data too well, including its noise and random fluctuations, so it performs poorly on new/unseen data. This condition can be recognized by the signs such as very low training error but high testing error.

Undefitting:- It refers to a model when it is too simple to capture the underlying pattern of the data, so it performs poorly on both training and testing data. It can be recognized by the signs of High error on training and testing sets.
For example:- Trying to fit a straight line to data that actually follows a curve; the model can't capture the trend.

Bias-Variance Trade-off:- It is defined as the balance between bias and variance that affects a model's performance:
* Bias:- Error due to overly simplistic assumptions (leads to underfitting).
* Variance:- Error due to too much senstivity to training data (leads to overfitting).

It's main aim is to find a model complex enough to capture patterns but simple enough to generalize well.




Ans 4: Outliers in a Dataset in Machine Learning are the deviations of the forecasted values from the standard values. Outliers are data points that are significantly different from the rest of the data. They can be unusually high or low compared to the majority of observations. Outiers can have a huge impact on the data as it can distort the statistics, affect the model performance and lead to misleading insights.

Three common techniques to handle the ouliers are:-
1. Removing Outliers:- It is a way by which you simply remove or delete the outlier data points if they are errors or irrelevant.

2. Transforming Data:- Apply transformations to reduce the effect of outliers (e.g., log transformation, square root, or Box-Cox).

3. Capping or Winsorization:- Replace extreme values with a maximum/minimum threshold (percentiles).

Ans 5: Handling missing values means applying methods to deal with gaps (NaN/NULL entries) in a dataset so the machine learning model can work properly. Missing values, if ignored, can reduce accuracy, bias results, or even prevent models from running.

Some methods to handle the missing errorsare :-
1. Identify:- This involves identifying missing values using methods like .isnull() in Python.
2. Analyze:- This involves analyzing the pattern to check if the data is missing randomly or systematically.
3. Decide:- This involves deciding the strategy, depending on the dataset either to remove rows or columns with too many missing or fill them using imputation.
4. Apply:- Applying the chosen method, i.e., Remove or Replace.

**Imputation Techniques for Categorial and Numerical data:-**

1. For Categorial data- Mode Imputation,
Replace missing numerical values with mean or median of that column

2. For Numerical data- Mean/Median Imputation, Replace missing categorical values with the most frequent category (mode).

```python
#Python Program to generate 1000 samples from a normal distribution
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Generate 1000 samples from a normal distribution
np.random.seed(42)  # for reproducibility
data = np.random.normal(loc=50, scale=10, size=1000)  # mean=50, std=10

# Create a DataFrame
df = pd.DataFrame(data, columns=['Value'])

# Show first 5 samples
print(df.head())

#Introduce 50 random missing values
# Step 2: Introduce 50 random missing values
np.random.seed(42)  # for reproducibility
missing_indices = np.random.choice(df.index, size=50, replace=False)
df.loc[missing_indices, 'Value'] = np.nan

# Check how many missing values we have
print("Number of missing values:", df['Value'].isna().sum())

#Fill missing values with column Mean
# Step 3: Fill missing values with the column mean
mean_value = df['Value'].mean()
df['Value'].fillna(mean_value, inplace=True)

# Verify that there are no missing values left
print("Number of missing values after imputation:", df['Value'].isna().sum())

#Plot a Histogram before and after imputation
# For plotting, create a copy of the original data with missing values
np.random.seed(42)
original_data = np.random.normal(loc=50, scale=10, size=1000)
original_df = pd.DataFrame(original_data, columns=['Value'])
missing_indices = np.random.choice(original_df.index, size=50, replace=False)
original_df.loc[missing_indices, 'Value'] = np.nan

# Plot histogram
plt.figure(figsize=(12,5))

# Before imputation
plt.subplot(1, 2, 1)
plt.hist(original_df['Value'].dropna(), bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram Before Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

# After imputation
plt.subplot(1, 2, 2)
plt.hist(df['Value'], bins=30, color='salmon', edgecolor='black')
plt.title('Histogram After Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
#Pyhton Program to implement Min-Max scaling on the following list of numbers [2, 5, 10, 15, 20] and print the scaled array.
from sklearn.preprocessing import MinMaxScaler
import numpy as np

# Original list of numbers
numbers = [2, 5, 10, 15, 20]

# Convert to a 2D array as required by sklearn
numbers_array = np.array(numbers).reshape(-1, 1)

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Fit and transform the data
scaled_numbers = scaler.fit_transform(numbers_array)

# Print the scaled array
print(scaled_numbers)
```
