Ans 1: Hypothesis testing is a statistical method used to make decisions or inferences about a population based on sample data. It helps us check whether a claim (assumption) about a population parameter is true or not, using sample evidence.
There are two types of hypothesis:
1. Null Hypothesis:- The "default" assumption, usually meaning "no effect" or "no difference."
2. Alternate Hypothesis:- What we want to test for; it contradicts the null.

Ans 2: The null hypothesis (H₀) is the starting assumption in hypothesis testing.
It usually states that there is no effect, no difference, or no relationship in the population. This is an assumption in which nothing new happens.
Fore example:-
  
  A coin is fair →
 H₀: The probability of heads = 0.5

The Alternate hypothesis is what we want to test or prove.It states that there is an effect, difference, or relationship in the population. It is an assumption in which something is happening.
For example:-

The coin is biased → H₁: Probability of heads ≠ 0.5

Ans 3: The significance level (α) is a threshold probability set by the researcher before conducting a hypothesis test. It represents the maximum risk of rejecting a true null hypothesis (Type I error) that you are willing to accept.
**Role of Significance Level in Hypothesis testing**:-
1. Defines the Rejection Region:- Based on α, we determine the critical value(s) of the test statistic. If the test statistic falls into the rejection region, we reject H₀.
2. Controls Type I error:- Type I Error = rejecting H₀ when it is actually true.
3. Comparison with P-value:- After calculating the p-value from sample data:
* If p ≤ α → Reject H₀ (evidence supports H₁).
* If p > α → Fail to reject H₀ (not enough evidence against H₀).


Ans 4: Type I errors are the false positive errors which occurs when we reject the null hypothesis (H₀) even though it is true. In other words, we detect an effect that does not actually exist whereas Type II errors occur when we fail to reject the null hypothesis (H₀) even though it is false. In other words, we miss a real effect that exists.

Example of a Type I Error:- A court trial: An innocent person is declared guilty based on some pbvious conditions that point towards him.
Example of Type II Error:- A court trial: A guilty person is declared innocent.

Ans 5: A Z-test and a T-test are both statistical tests used in hypothesis testing, but they differ mainly in the assumptions about the population and the sample size.

Z-Test:- A Z-test is used when the population variance (σ²) or standard deviation (σ) is known and the sample size is large (typically n > 30). It relies on the normal distribution. For example, if you know the population standard deviation of exam scores and want to test whether a sample mean differs from the population mean, you would use a Z-test.

T-Test:- A T-test, on the other hand, is used when the population variance is unknown and must be estimated from the sample. It is especially useful when the sample size is small (n ≤ 30). The test statistic follows the t-distribution, which has heavier tails than the normal distribution to account for the extra uncertainty from estimating the population standard deviation. For instance, if you only have a small class sample and do not know the population standard deviation, you would use a T-test to compare the sample mean with the hypothesized population mean.

We use a Z-test mainly when the population standard deviation is known and the sample size is large (n > 30) and the T-test when the population standard deviation is unknown and you estimate it from the sample, especially with small samples (n ≤ 30).

```python
#Python Program to generate a binomial distribution with n=10 and p=0.5 and to then Plot a Histogram
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10      # number of trials
p = 0.5     # probability of success
size = 1000 # number of samples

# Generate binomial distribution
data = np.random.binomial(n, p, size)

# Print first 20 values
print("Sample Binomial Distribution Data:", data[:20])

# Plot histogram
plt.hist(data, bins=range(n+2), edgecolor='black', alpha=0.7)
plt.title(f"Binomial Distribution (n={n}, p={p})")
plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.show()
```

```python
#Python Program to implement the Hypothesis Testing using Z-statistics for a sample dataset
import numpy as np
from scipy.stats import norm

# Sample data
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Hypothesized population mean (H0)
mu_0 = 50

# Known population standard deviation (assumption for Z-test)
sigma = 1

# Calculate sample mean and sample size
sample_mean = np.mean(sample_data)
n = len(sample_data)

# Compute Z-statistic
z_stat = (sample_mean - mu_0) / (sigma / np.sqrt(n))

# Calculate p-value (two-tailed test)
p_value = 2 * (1 - norm.cdf(abs(z_stat)))

# Significance level
alpha = 0.05

# Print results
print("Sample Mean:", sample_mean)
print("Z-statistic:", z_stat)
print("p-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject the null hypothesis (H0).")
else:
    print("Fail to reject the null hypothesis (H0).")
```

```python
#Python code to  simulate data from a normal distribution and calculate the 95% confidence interval for its mean, also plot the data using Matplotlib.
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Simulate data from a normal distribution
np.random.seed(42)  # for reproducibility
mu, sigma = 50, 5   # population mean and std deviation
n = 100             # sample size
data = np.random.normal(mu, sigma, n)

# Calculate sample mean and standard error
sample_mean = np.mean(data)
std_err = stats.sem(data)  # standard error of the mean

# 95% confidence interval
confidence = 0.95
ci = stats.t.interval(confidence, df=n-1, loc=sample_mean, scale=std_err)

print("Sample Mean:", sample_mean)
print("95% Confidence Interval:", ci)

# Plot the data
plt.hist(data, bins=15, edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2, label=f"Mean = {sample_mean:.2f}")
plt.axvline(ci[0], color='green', linestyle='dotted', linewidth=2, label=f"Lower CI = {ci[0]:.2f}")
plt.axvline(ci[1], color='green', linestyle='dotted', linewidth=2, label=f"Upper CI = {ci[1]:.2f}")
plt.title("Normal Distribution with 95% Confidence Interval")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
#Python Function to calculate the Z-scores from thhe dataset and visualize the standardized data using a histogram
import numpy as np
import matplotlib.pyplot as plt

# Function to calculate Z-scores
def calculate_z_scores(data):
    mean = np.mean(data)
    std = np.std(data)
    z_scores = [(x - mean) / std for x in data]
    return np.array(z_scores)

# Sample dataset
data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
        50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
        50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Calculate Z-scores
z_scores = calculate_z_scores(data)

# Print first 10 Z-scores
print("First 10 Z-scores:", z_scores[:10])

# Plot histogram of Z-scores
plt.hist(z_scores, bins=10, edgecolor='black', alpha=0.7)
plt.title("Histogram of Standardized Data (Z-scores)")
plt.xlabel("Z-score")
plt.ylabel("Frequency")
plt.axvline(0, color='red', linestyle='dashed', linewidth=2, label="Mean = 0")
plt.legend()
plt.show()
```

Z-scores tell you how many standard deviations a data point is away from the mean of the dataset.
* A Z-score of 0 means the value is exactly equal to the mean.
* A positive Z-score means the value is above the mean.
* A negative Z-score means the value is below the mean.

In our dataset,
* the mean is around 50
* If a data point like 51.0 has a Z ≈ +1, it means it's about 1 std dev above average
* If a data point like 48.7 has a Z ≈ -1, it means it's about 1 std dev below average
So, Z-scores standardize the data, making it easier to compare values relative to the overall distribution, regardless of the original units.
