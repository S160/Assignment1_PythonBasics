Ans 2: Broadcasting is a set of rules that NumPy follows to perform arithmetic operations on arrays of different shapes and sizes—without explicitly copying data. It's like NumPy “stretches” smaller arrays across larger ones so they match in shape and the operation can be done element-wise.

When NumPy operates on two arrays:

1. It compares shapes from right to left (trailing dimensions)
2. Two-dimensions are compatible if:
  *  They are equal, or
  *  One of them is 1 ('started').
3. If they are incompatible in any dimension, NumPy throws an error.

Broadcasting is useful because it avoids explicit loops, so its faster. It also saves memory as there is no duplication of files or data. It makes code concise and readable.

NumPy auto-expands smaller arrays to match the bigger arrays without actually copying their data.

Ans 21: NumPy array slicing is faster than Python list slicing due to the following:-

**Python Lists**:-
* A Python list is a collection of pointers to objects stored in memory.
* Each element can be of a different type.
* When you slice a list, Python has to create a new list and copy pointers for each element.
* Python list slicing always makes a new copy of the sliced portion.

**NumPy Arrays**:-
* NumPy arrays (ndarray) are stored in contiguous blocks of memory.
* All elements are of the same data type (homogeneous).
* When you slice a NumPy array, it creates a view (not a copy) of the same memory buffer → much faster.
* NumPy slicing returns a view of the original array (no copying of data).The sliced array just references the same memory with a different offset.

```python
#Python Program to calculate the correlation matrix using Seaborn and visualize it with a heatmap
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample DataFrame
data = {
    "Maths": [85, 78, 92, 70, 88, 95],
    "Physics": [80, 75, 89, 68, 85, 90],
    "Chemistry": [78, 82, 91, 65, 84, 88],
    "Biology": [72, 76, 85, 60, 80, 86]
}

df = pd.DataFrame(data)

# Calculate correlation matrix
corr_matrix = df.corr()

print("Correlation Matrix:")
print(corr_matrix)

# Plot heatmap
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
