Ans 1: The difference between interpreted and compiled languages lies in how the source code is translated into machine code and executed:

In Compiled Languages the source code is translated entirely into machine code by a compiler before execution whereas in Interpreted Languages the source code is read and executed line by line by an interpreter at runtime.

The execution process in case of Compiled Languages involves:

a) Writing the source code

b) Compiling it into an executable file and then

c) Run the executable directly.

Example:- C, C++, Rust, Go.

Whereas, in case of Interpreted languages, execution involves:

a) Writing the source code and

b) Running it directly in an interpreter (no compilation)

Example:- Python, JavaScript, Ruby.

Ans 7: In Python, multiple exceptions can be handled so that your program can respond differently to different types of errors or handle several error possibilities in one block.

1. Multiple except Blocks:- You write a separate except block for each type of exception. This allows you to give a specific response for each error type. You get a clear and specific error handling for each case.

  3. Using Exception to handle all Errors (Generic):- You can catch any type of exception by catching the base Exception class. This is less specific, so it might hide unexpected bugs. Best used for logging or final safety.

```python
try:
    file = open("nonexistent.txt")
except Exception as e:
    print(f"An error occurred: {e}")
```

Ans 11: Memory Management in Python refers to the process of efficiently allocating, using, and freeing memory during the execution of a Python program.

Memory Management in Python involves:-

1. Automatic Memory Allocation:- Python automatically handles the allocation of memory when you create variables, objects, or data structures. This is done by Python's memory manager, so you don't need to manually allocate memory (as in languages like C).
2. Garbage Collection:- Python automatically frees memory that is no longer in use through garbage collection. The most common method is reference counting:
Every object has a count of references pointing to it. When the count drops to zero (meaning no variable is referencing it), the memory is released. Python also uses a cyclic garbage collector to handle reference cycles.
3. Private Heap Space:- All Python objects and data structures are stored in a private heap space. The memory manager allocates space from this heap and ensures efficient use.
4. Built-in Functions for Memory Control:- gc module allows manual control over garbage collection. sys.getsizeof() can be used to check the size of an object in bytes.

Ans 15: Python's garbage collection system automatically reclaims memory by removing objects that are no longer in use, so the programmer doesn’t have to manually free memory (like in C/C++).
It works by:
1. Reference Counting:- Every object in Python keeps track of how many references point to it. This count increases when you assign the object to a variable or pass it as a parameter, and decreases when those references go out of scope or are deleted. When the reference count drops to zero, Python immediately frees the memory for that object.
2. Garbage Collector for Cyclic References:- Reference counting can't handle circular references alone i.e., 2 or more objects reference each other but are no longer accessible.

    So, the Python's garbage collector scans them and frees them. It uses a generalized collection strategy:-

    a) Classifies objects into generations (young, middle-aged, old)

    b) Younger generations are scanned more often beacause they more likely to become unused.
3. Finalization (__del__ method):- f an object defines a __del__ method (destructor), the garbage collector calls it before freeing memory.
However, having __del__ in cyclic references can complicate cleanup, so Python may not collect such cycles immediately.

```python
#Program in Python to implement a basic logging setup that logs to a file with rotation after 1MB
import logging
from logging.handlers import RotatingFileHandler

# Configure logger
logger = logging.getLogger("MyLogger")
logger.setLevel(logging.DEBUG)  # Capture all levels of logs

# Create a rotating file handler
handler = RotatingFileHandler(
    "app.log",       # Log file name
    maxBytes=1 * 1024 * 1024,  # 1 MB
    backupCount=3    # Keep last 3 backup files
)

# Create log format
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
handler.setFormatter(formatter)

# Add handler to logger
logger.addHandler(handler)

# Example log messages
for i in range(2000):
    logger.debug(f"Debug message {i}")
    logger.info(f"Info message {i}")
    logger.warning(f"Warning message {i}")
    logger.error(f"Error message {i}")
    logger.critical(f"Critical message {i}")

print("Logging complete. Check 'app.log' and rotated log files.")
```
