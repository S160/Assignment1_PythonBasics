Ans 1: SLR is a statistical method used to study the relationship between two variables:- one independent variable (X) and one dependent (Y). Simple Linear Regression models the relationship between X and Y by fitting a straight line (called the regression line) to the observed data.

The equation of this line is:

        Y=β0​+β1​X+ε

Where,
 X- the independent variable (the outcome or response we want to predict)

 Y- the dependent variable (the predictor or input variable)

 β₀- Intercept (value of Y when X = 0)

 β₁- Slope (the amount Y changes for each unit increase in X)

 ε-  Error term (represents random variation not explained by X).

 The main purpose of SLR:-
1. Prediction: To predict the value of the dependent variable (Y) based on the known value of the independent variable (X).
For example:- Predicting the student's marks (Y) based on the no. of hours studied (X).
2. Relationship Analysis: To determine whether and how strongly the two variables are related (direction and strength of association).
For example: Analyzing if advertising expenditure (X) significantly affects sales revenue (Y).
3. Quantifying Impact: To estimate how much Y changes when X increases by one unit — given by the slope (β₁).
4. Trend Estimation: To identify and model trends in data where one variable tends to increase or decrease linearly with another.

Ans 2: Key assumptions of SLR are:-

1. Linearity:- The relationship between the independent variable (X) and the dependent variable (Y) is linear.

Mathematically, it is:

               Y=β0​+β1​X+ε
2. Independence of  Errors:- The residuals (errors, ε) are independent of each other. This means the error for one observation does not depend on the error for another.
3. Homoscedasticity (Constant Variance of Errors): The variance of residuals is constant across all levels of X. In other words, the spread of errors should be roughly the same for all predicted values. If residuals fan out or narrow down → heteroscedasticity (violates the assumption).
4. Normality of Errors: The residuals (ε) are normally distributed. This is especially important for making statistical inferences (like confidence intervals and hypothesis tests).
5. No (or Minimal) Measurement Error in X:- The independent variable (X) should be measured accurately.If X contains large measurement errors, it can bias the regression estimates.

Ans 3: The mathematical equation for a Simple Linear Regression (SLR) model is:

      𝑌= 𝛽0+𝛽1𝑋+𝜀

where,

Y- (Dependent/Response variable), The outcome we are trying to predict or explain.

X- (Independent/Predictor Variable) The input variable used to make predictions about Y.

𝛽0- (Intercept) The value of 𝑌 when 𝑋=0. It represents where the regression line crosses the Y-axis.

𝛽1- (Slope Coefficient) The change in 𝑌 for a one-unit increase in X. It shows the strength and direction of the relationship between X and Y.

𝜀- (Error/ Residual Term) The difference between the observed value and the predicted value. It captures variability not explained by X.

Ans 5: The Method of Least Squares is the most commonly used technique to estimate the best-fitting line in linear regression. Its goal is to find a regression line that minimizes the total error between the observed data points and the predicted values.

The regression model predicts a value for each observed data point. The difference between the actual value (Y) and the predicted value (Ŷ) is called the residual (error):

      Residual=Y−Y^
The line for which the sum of squared residuals is as small as possible.

      Minimize ∑(Y−Y^)2
We square the errors to ensure:
1. Negative and Positive errors do not cancel out
2. Large errors are penalized more

The Regression Line in SLR, the fitted line equation is:

        Y^=β0​+β1​X
The method of least squares finds the best values of:
1. 𝛽0 (intercept)
2. 𝛽1 (slope)

that minimizes  ∑(Y−Y^)2

The method of Least Squares in Simple Linear Regression gives the most accurate linear line for the data. It is easy to compute and interpret. It works well even for large datasets.

Ans 6: Logistic Regression is a statiscal & Machine Learning classification Algorithm used to predict the probability of a categorial outcome most commonly binary outcomes (such as Yes/No, spam/not spam, 0/1).

Logistic Regression mathematically does not fit as a straight line like Linear Regression, rather it uses the Sigmoid function to ensure the output is always between 0 and 1.

**Mathematical form (Sigmoid Function)**

      P(Y=1∣X)= 1/[1+e^−(β0+β1X)]

*  Output is a probability between 0 & 1
*  Based on a threshold (mainly 0.5) it is classified into classes (0,1)

Logistic Regression is different from Linear Regression as it predicts a categorial outcome or probability whereas Linear Regression predicts a continuous value.

The output range of Linear Regression is from (−∞ to +∞) whereas in Logistic Regression the output range varies from 0 to 1.

The straight line equation used in Linear Regression is:

       Y=β0​+β1​X+ε
whereas the equation of Sigmoid probability function P, in Logistic Regression is:
      
      P(Y=1∣X)= 1/[1+e^−(β0+β1X)]

Linear Regression uses 'Regrssion' model for prediction of numeric values whereas Logistic Regression uses the 'Classification' model for class labels.

The best-fit method used in Linear Regresion is the OLS (Ordinary Least Square) method & Logistic Regression uses the MLE (Maximum Likelihood Estimation) method as its best-fit method.

In simple words, if Linear Regression predicts how much? (price, sales) then Logistic Regression predicts which class? (yes/no, 0/1)

For example:- In Logistic Regression:-
* for predicting whether an E-mail is a spam (1) or not (0).
* Predicting sales, whether a customer will buy (1) or not (0)

1. **Mean Absolute Error (MAE)**:- The average of the absolute differences between the actual and the predicted values is called the Mean Absolute Error or(MAE). It measures how far the predictions made are from the true values on an average.

      Lower MAE indicates that it is a better model. The equation is:-

        MAE=1/n(1​∑∣yi​−y^​i​∣)
  
2. **Mean Squared Error (MSE)**:- The average of the squared differences between the actual and predicted values is called Mean Squared Error or (MSE). It penalizes larger errors more than MAE (because of the square).

        MSE=1/n(∑(yi​−y^​i​)2)

    R^2 ranges from 0 to 1:
    * It is closer to 1, better fit
    * 0 means the model explains everything

Ans 8: The R-squared (R²) metric in regression analysis measures how well the independent variables explain the variability of the dependent variable. In simpler terms,
1. It tells you the proportion of the total variation in the dependent variable (Y) that is explained by the regression model.
2. Range 0 to 1:
    * 0: The model explains none of the variability in Y
    * 1: The model explains all the variablitiy in Y.

        
            
          R² = 1-(SSR/SST)= 1−(∑(yi−y^i)^2/∑(yi​−yˉ​​)2​)
	​
Where, yi:- actual value

y^i:- predicted value

  yˉ:- mean of actual values

Higher the R², the better the model fits the data & lower the R², the model does not explains much of the variation in Y.

```python
#Python program for fitting a Simple Linear Regression model using scikit-learn and printing the slope and intercept:
# Import necessary libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Sample data
# X = independent variable, Y = dependent variable
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # reshape for sklearn
Y = np.array([2, 4, 5, 4, 5])

# Create the model
model = LinearRegression()

# Fit the model
model.fit(X, Y)

# Print slope (coefficient) and intercept
print("Slope (b1):", model.coef_[0])
print("Intercept (b0):", model.intercept_)
```
