```python
#Python code to plot the distribution of selling prices using a histogram
#import the required libraries
import pandas as pd
import matplotlib as plt

#Load dataset
bike_df=pd.read_csv("/mnt/data/BIKE DETAILS.csv")

#Plot histogram for selling_price
plt.figure(figsize=(8,5))
plt.hist(bike_df['selling price'], bins=30, edgecolor='black')
plt.title('Distribution of Selling Prices')
plt.xlabel(Selling price($))
plt.ylabel('Frequency')
plt.grid[axis='y', linestyle= '--', alpha-0.7]
plt.show
```

This histogram displays how frequently bikes are sold across various price ranges.

The curve is right-skewed showing:
* Most bikes are sold at lower prices or mid-range prices (around 20000-100000)
* A small number of bikes have very high selling prices which stretches the distribution tail.

This indicates that affordable bikes dominate the resale market while premium or superbikes are comparatively rare.

```python
#Python code to plot a bar plot to visualize the average selling prices for each seller type & write one observation
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
bike_df = pd.read_csv("/mnt/data/BIKE DETAILS.csv")

# Calculate average selling price for each seller type
avg_price_by_seller = bike_df.groupby('seller_type')['selling_price'].mean()

# Create bar plot
plt.figure(figsize=(7,5))
avg_price_by_seller.plot(kind='bar', color=['skyblue', 'lightgreen'], edgecolor='black')
plt.title('Average Selling Price by Seller Type')
plt.xlabel('Seller Type')
plt.ylabel('Average Selling Price (₹)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

The average selling price of bikes sold by individuals is higher than that of dealers.
This suggests that individual sellers may offer better-maintained or premium bikes, while dealers tend to sell bikes at lower prices, possibly to clear inventory or attract more buyers.

```python
#Python code to compute the average km_driven for each ownership type(1st owner, 2nd owner, etc.)
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
bike_df = pd.read_csv("/mnt/data/BIKE DETAILS.csv")

# Calculate average km_driven for each ownership type
avg_km_by_owner = bike_df.groupby('owner')['km_driven'].mean().sort_values()

# Create bar plot
plt.figure(figsize=(8,5))
avg_km_by_owner.plot(kind='bar', color='lightcoral', edgecolor='black')
plt.title('Average Kilometers Driven by Ownership Type')
plt.xlabel('Ownership Type')
plt.ylabel('Average Kilometers Driven')
plt.xticks(rotation=30)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
#Python code for detecting and removing outliers from km_driven column
# Import required libraries
import pandas as pd

# Load dataset
bike_df = pd.read_csv("/mnt/data/BIKE DETAILS.csv")

# --- Before removing outliers ---
before_summary = bike_df['km_driven'].describe()

# --- IQR Method ---
Q1 = bike_df['km_driven'].quantile(0.25)
Q3 = bike_df['km_driven'].quantile(0.75)
IQR = Q3 - Q1

# Define bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Remove outliers
bike_df_no_outliers = bike_df[(bike_df['km_driven'] >= lower_bound) & (bike_df['km_driven'] <= upper_bound)]

# --- After removing outliers ---
after_summary = bike_df_no_outliers['km_driven'].describe()

print("Before Removing Outliers:\n", before_summary)
print("\nAfter Removing Outliers:\n", after_summary)
```

```python
#Python code to Generate a heatmap of the correlation matrix for all numeric columns.
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
bike_df = pd.read_csv("BIKE DETAILS.csv")

# Compute correlation matrix (only for numeric columns)
corr_matrix = bike_df.corr(numeric_only=True)

# Plot the heatmap
plt.figure(figsize=(8, 5))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap for Numeric Columns')
plt.show()
```

The correlations that sand out the most are as follows:
1. year has a positive correlation with selling_price, meaning newer bikes tend to have higher prices.

2. km_driven has a negative correlation with selling_price, suggesting that more-used bikes usually sell for less.

3. year and km_driven often show a mild negative correlation, as older bikes tend to have more kilometers driven.
