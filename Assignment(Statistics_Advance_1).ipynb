Ans 5: The standard normal distribution is a special case of the normal distribution (bell-shaped curve).
It has:
1. Mean (μ) = 0
2. Standard deviation (σ) = 1

The random variable that follows this distribution is usually written as Z.
It is important because of the following reasons:

1. It is the basis of the Normal family. Any normal distribution
𝑋∼𝑁 (𝜇,𝜎)X~N(μ,σ^2) can be converted into a standard normal using the Z-score transformation:
              
               Z=(𝑋-𝜇)/𝜎
2. Z-scores:- A Z-score tells you how many standard deviations a value is away from the mean
3. Probability Calculations:- Many statistical methods (hypothesis testing, confidence intervals, control charts, etc.) rely on standard normal probabilities. Instead of recalculating for each normal distribution, we use standard normal tables (or software).
4. Central Limit Theorem (CLT):- The CLT says: for large sample sizes, the sample mean tends to follow a normal distribution, often standardized to the standard normal.

Ans 6: The Central Limit Theorem (CLT) states that If you take a large number of independent random samples from any population (with finite mean 𝜇
μ and finite variance 𝜎^2), then the sampling distribution of the sample mean will be approximately normal, regardless of the original population's shape.

          Xˉ∼N(μ,σ^2/n​)    as n→∞

Where:

1. X = sample mean
2. μ = population mean
3. 𝜎^2= population variance
4. n = sample size

In Statistics, CLT is significant because:

1. Normal Approximation:- Even if the population is skewed or not normal, the distribution of sample means tends to be normal if n is large enough. This is why the normal distribution appears so often in real life.
2. Foundation for Inference:- Hypothesis testing, p-values, and confidence intervals rely on the CLT. It allows us to make probability-based statements about sample statistics.
3. Standard Error:-The variability of the sample mean decreases as sample size increases:
        
        SE= ​σ/(​n)^1/2
This explains why larger samples give more reliable estimates.

Ans 7: A confidence interval gives a range of values that is likely to contain the true population parameter (like the mean, proportion, or difference) with a certain level of confidence (commonly 95% or 99%).

The significance of confidence intervals in statistical analysis is:

1. Quantifies Uncertainty:- A single point estimate (like a sample mean = 170 cm) doesn’t show uncertainty. A CI shows the margin of error around the estimate.
2. Confidence Intervals are more informative than Hypothesis Tests. Hypothesis testing gives a binary decision (reject/not reject). A CI gives a range of plausible values, which is more insightful.
3. Directly Interpretable:- For example, if a 95% CI for drug effectiveness is (2%, 8%), it suggests the drug improves outcomes by between 2% and 8%.
4. Helps Assess Precision:- Narrow CI → estimate is precise (large sample, low variability). Wide CI → estimate is uncertain (small sample, high variability).

Ans 8: The expected value of a random variable is the long-run average outcome you would expect if you repeated an experiment many times.
It’s essentially the “center of gravity” or weighted average of the distribution.

```python
#Python Program to generate 1000 random numbers from a normal distribution with mean = 50 and standard deviation = 5.
#Compute its mean and standard deviation using NumPy, and draw a histogram to visualize the distribution.
import numpy as np
import matplotlib.pyplot as plt

# Parameters
mean = 50
std_dev = 5
n_samples = 1000

# Generate random numbers from normal distribution
data = np.random.normal(loc=mean, scale=std_dev, size=n_samples)

# Compute mean and standard deviation
sample_mean = np.mean(data)
sample_std = np.std(data)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Sample Standard Deviation: {sample_std:.2f}")

# Plot histogram
plt.hist(data, bins=30, edgecolor='black', alpha=0.7)
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2,_
```

Ans 10: Here’s how you’d apply it:

1. Collect a sample of daily sales (or use the full dataset as one sample).

2. Compute the sample mean (𝑋ˉ) and sample standard deviation (s).

3. Compute the Standard Error (SE):

      
      𝑆𝐸= 𝑠/(𝑛)^1/2
where
n = sample size.

Construct the 95% Confidence Interval:
Since the CLT allows us to assume normality of the sample mean:

         𝐶𝐼=𝑋ˉ±𝑍𝛼/2⋅𝑆𝐸


𝑍𝛼/2 = 1.96 for a 95% confidence interval
𝑋ˉ= sample mean
SE = standard error

This gives a range in which the true average daily sales likely falls.

```python
import numpy as np
import scipy.stats as stats

# Daily sales data
daily_sales = [220, 245, 210, 265, 230, 250, 260, 275, 240, 255,
               235, 260, 245, 250, 225, 270, 265, 255, 250, 260]

# Convert to NumPy array
data = np.array(daily_sales)

# Compute mean
mean_sales = np.mean(data)

# Sample size, standard error
n = len(data)
std_err = stats.sem(data)

# 95% confidence interval
confidence = 0.95
ci = stats.t.interval(confidence, df=n-1, loc=mean_sales, scale=std_err)

print(f"Mean Sales: {mean_sales:.2f}")
print(f"95% Confidence Interval: {ci}")
```
